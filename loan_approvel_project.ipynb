```python
import numpy as np 
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

working with missing values


```python
# we have numrical mising values and  catogorical missing values 
object_cols =df2.select_dtypes(include="object")
object_cols

numerical_cols =df2.select_dtypes(include ="number")
numerical_cols

for cols in numerical_cols:
    df2[cols] = df2[cols].fillna(df2[cols].mean())


for cols in object_cols:
    df2[cols] = df2[cols].fillna(df2[cols].mode()[0])

df2.isnull().sum()


#see i have cleared all the missing values ok 
```

```python
df2
```

```python
#see we have application id thats notrequired  ok now we drop the items or the rows thats not required 

df2 =df2.drop("Applicant_ID",axis =1)
```
