```python
import numpy as np
```

```python
class LogisticRegression:

    def __init__(self,learning_rate =0.1,iters =1000):
        self.bias =None
        self.weights =None
        self.lr =learning_rate
        self.num_iter =iters
        
    def _sigmoid(z):
        return (1/(1 + np.exp(-z)))
        
    def fit(self,X,y):
        m,n =X.shape
        self.bias =0
        self.weights =np.zeros(n)

        for i in range(self.num_iter):
            z = self.bias + np.dot(X.T ,self.weights)
            y_pred =self._sigmoid(z)

            db = (1/m) * (np.sum(y_pred -y))
            dw = (1/m) *(np.dot(X.T, y_pred  - y))

            self.bias -= lr *db
            self.weights -=lr *dw
            

            
```
